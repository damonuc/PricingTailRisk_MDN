```python
import os
os.chdir('/Users/jianbinchen/NonSync/GitHub/Research/PricingTailRisk_MDN/overleaf/')

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.integrate import quad
from scipy.optimize import fsolve

# 1. 基础参数
params = [(-0.35, 0.3, 0.3), (0.1, 0.15, 0.4), (0.3, 0.3, 0.3)]
# 【修改1】将绘图用到的 x 范围改小，保证在 -1 到 1 之间有更高的分辨率 (两端稍微留点余量到-2和2防止边缘截断不平滑)
x = np.linspace(-2, 2, 1000)

# 2. 定义混合分布的 PDF 和 CDF
def mixture_pdf(x_val):
    return sum(w * norm.pdf(x_val, m, s) for m, s, w in params)

def mixture_cdf(x_val):
    return sum(w * norm.cdf(x_val, m, s) for m, s, w in params)

# 计算混合分布的整体均值 (mu) 和 标准差 (sigma)
mix_mu = sum(w * m for m, s, w in params)
mix_var = sum(w * (s**2 + m**2) for m, s, w in params) - mix_mu**2
mix_sigma = np.sqrt(mix_var)

# 1. 先计算等效正态分布的 VaR，把它作为寻找真实 VaR 的绝佳起点
z_5pct = norm.ppf(0.05)
guess_var = mix_mu + mix_sigma * z_5pct

# 2. 将 guess_var 传入 fsolve
var_5pct_mix = fsolve(lambda x: mixture_cdf(x) - 0.05, guess_var)[0]
es_res_mix, _ = quad(lambda x: x * mixture_pdf(x), -20, var_5pct_mix)
es_5pct_mix = es_res_mix / 0.05

# 4. 计算等效 Normal 的 VaR 和 ES (5%)
z_5pct = norm.ppf(0.05)
var_5pct_norm = mix_mu + mix_sigma * z_5pct
es_5pct_norm = mix_mu - mix_sigma * (norm.pdf(z_5pct) / 0.05)

# --- 开始绘图 ---
plt.figure(figsize=(12, 8))
y_mixture = [mixture_pdf(val) for val in x]
y_normal = norm.pdf(x, mix_mu, mix_sigma)

# 为 3 个子分布赋予不同颜色
comp_colors = ['#2ca02c', '#ff7f0e', '#9467bd'] # 绿, 橙, 紫
for i, (m, s, w) in enumerate(params):
    plt.plot(x, w * norm.pdf(x, m, s), ':', color=comp_colors[i], alpha=0.7, label=f'Component {i+1}')

# 画出主要的两个密度曲线
plt.plot(x, y_mixture, color='black', lw=2.5, label='Mixture Density')
plt.plot(x, y_normal, color='royalblue', lw=2.5, linestyle='--', label='Equivalent Normal')

# 尾部风险区域 (左侧 5%) - ES 的区域表示
x_tail_mix = np.linspace(-10, var_5pct_mix, 200)
plt.fill_between(x_tail_mix, [mixture_pdf(v) for v in x_tail_mix], color='red', alpha=0.3, label='Mixture 5% ES Area')

x_tail_norm = np.linspace(-10, var_5pct_norm, 200)
plt.fill_between(x_tail_norm, norm.pdf(x_tail_norm, mix_mu, mix_sigma), color='royalblue', alpha=0.3, label='Normal 5% ES Area')

# 背景填充 (可选)
plt.fill_between(x, y_mixture, color='gray', alpha=0.05)

# 仅保留均值和 VaR 的垂线，删除 ES 垂线
plt.axvline(mix_mu, color='black', linestyle=':', lw=1.5, alpha=0.6)
plt.axvline(var_5pct_mix, color='darkred', linestyle='-', lw=1.5, label=f'Mix VaR: {var_5pct_mix:.2f}') 
plt.axvline(var_5pct_norm, color='darkblue', linestyle='--', lw=1.5, label=f'Norm VaR: {var_5pct_norm:.2f}')

# 文本框标注统计量
stats_text = (
    f"--- Shared Stats ---\n"
    f"$\mu$     = {mix_mu:>6.2f}\n"
    f"$\sigma$     = {mix_sigma:>6.2f}\n\n"
    f"--- 5% Tail Risk ---\n"
    f"Mix VaR = {var_5pct_mix:>6.2f}\n"
    f"Nor VaR = {var_5pct_norm:>6.2f}\n"
    f"Mix ES  = {es_5pct_mix:>6.2f}\n"
    f"Nor ES  = {es_5pct_norm:>6.2f}"
)
# 【修改2】调整文本框 x 轴坐标，从 -9.5 改为 -0.95，防止超出新的显示范围
plt.text(-0.95, 1, stats_text, bbox=dict(facecolor='white', alpha=0.9, edgecolor='lightgray'), 
         fontsize=12, family='monospace', verticalalignment='top')

# 格式化
plt.title('Tail Risk Comparison: Mixture Normal vs. Equivalent Normal (5%)', fontsize=15, pad=15)
plt.xlabel('Returns', fontsize=12)
plt.ylabel('Density', fontsize=12)

# 图例布局优化
plt.legend(loc='upper right', prop={'family': 'monospace', 'size': 10}, framealpha=0.9)
# 【修改3】将 X 轴范围限制为 -1 到 1
plt.xlim(-1, 1)
plt.ylim(bottom=0) # 让 y 轴从 0 开始，看起来更扎实
plt.grid(axis='y', alpha=0.3, linestyle='--')

plt.savefig('./fig/mixture_vs_normal_tail_risk_clean.pdf', bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.integrate import quad
from scipy.optimize import fsolve

# 1. 基础参数
params = [(-3, 1, 0.1), (-1, 0.5, 0.1), (0, 0.8, 0.3), (1.5, 0.6, 0.3), (4, 1.2, 0.2)]
x = np.linspace(-10, 10, 1000)

# 2. 定义混合分布的 PDF 和 CDF
def mixture_pdf(x_val):
    return sum(w * norm.pdf(x_val, m, s) for m, s, w in params)

def mixture_cdf(x_val):
    return sum(w * norm.cdf(x_val, m, s) for m, s, w in params)

# 计算混合分布的整体均值 (mu) 和 标准差 (sigma)
mix_mu = sum(w * m for m, s, w in params)
# 混合分布方差公式: E[X^2] - (E[X])^2
mix_var = sum(w * (s**2 + m**2) for m, s, w in params) - mix_mu**2
mix_sigma = np.sqrt(mix_var)

# 3. 数值计算 5% VaR
# VaR 是满足 CDF(x) = 0.05 的那个 x
var_5pct = fsolve(lambda x: mixture_cdf(x) - 0.05, -5)[0]

# 4. 数值计算 5% Expected Shortfall (ES)
# ES = (1/alpha) * integral from -inf to VaR of (x * pdf)
es_res, _ = quad(lambda x: x * mixture_pdf(x), -20, var_5pct)
es_5pct = es_res / 0.05

# --- 开始绘图 ---
plt.figure(figsize=(11, 7))
y_mixture = [mixture_pdf(val) for val in x]

# 画出单个成分
for m, s, w in params:
    # plt.plot(x, w * norm.pdf(x, m, s), '--', alpha=0.3, lw=1)
    plt.plot(x, w * norm.pdf(x, m, s), '--', alpha=0.5, label=f'μ={m:>4.1f}, σ={s:>4.1f}, w={w:>4.1f}') # 画出单个成分

# 画出混合曲线
plt.plot(x, y_mixture, color='black', lw=2.5, label='Mixture Density')

# 5. 标注 VaR 区域 (左侧 5% 尾部)
x_tail = np.linspace(-10, var_5pct, 200)
plt.fill_between(x, y_mixture, color='gray', alpha=0.2)
plt.fill_between(x_tail, [mixture_pdf(v) for v in x_tail], color='red', alpha=0.3, label='5% Risk Area')

# 6. 画出 VaR, ES 和 Mu 的垂直线
plt.axvline(var_5pct, color='red', linestyle='-', lw=1.5, label=f'VaR(5%): {var_5pct:.2f}') 
# plt.axvline(es_5pct, color='darkred', linestyle='-.', lw=1.5, label=f'ES(5%): {es_5pct:.2f}')
plt.axvline(mix_mu, color='blue', linestyle=':', lw=1.5, label=f'$\mu_{{mix}}$: {mix_mu:.2f}')

# 文本框标注统计量
stats_text = (f'Mixture Stats:\n'
              f'$\mu = {mix_mu:.2f}$\n'
              f'$\sigma = {mix_sigma:.2f}$\n'
              f'$VaR_{{5\%}} = {var_5pct:.2f}$\n'
              f'$ES_{{5\%}} = {es_5pct:.2f}$')
plt.text(-9.5, 0.15, stats_text, bbox=dict(facecolor='white', alpha=0.8), fontsize=14, family='monospace')

# 格式化
plt.title('Mixture Normal Density with VaR & ES (5%)', fontsize=14)
plt.xlabel('Returns / Value')
plt.ylabel('Density')
plt.legend(loc='upper right', prop={'family': 'monospace', 'size': 10})
plt.grid(axis='y', alpha=0.3)

# 保存
plt.savefig('./fig/mixture_density_example.pdf',  bbox_inches='tight')
plt.show()
```
